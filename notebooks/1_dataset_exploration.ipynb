```python
import torch
from torch_geometric.data import Data

def create_graph(row, label):
    x = torch.tensor(
        row,
        dtype=torch.float32
    ).view(31, 1)

    y = torch.tensor(
        label,
        dtype=torch.long
    )

    return Data(
        x=x,
        edge_index=edge_index,
        y=y
    )
```

```python
def create_graph_dataset(X, y):
    graphs = []

    for i in range(len(X)):
        graph = create_graph(X[i], y.iloc[i])
        graphs.append(graph)

    return graphs
```

```python
import torch
import torch.nn as nn

from torch_geometric.nn import GCNConv, global_mean_pool
```

```python
class ServiceGuardGNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv1 = GCNConv(1, 32)
        self.conv2 = GCNConv(32, 64)

        self.classifier = nn.Linear(64, 32)

    def forward(self, x, edge_index, batch):

        x = self.conv1(x, edge_index)
        x = torch.relu(x)

        x = self.conv2(x, edge_index)
        x = torch.relu(x)

        x = global_mean_pool(x, batch)

        x = self.classifier(x)

        return x
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
# 1. Get batch
# 2. Clear old gradients
# 3. Forward pass
# 4. Calculate loss
# 5. Backpropagation
# 6. Update weights

model.train()

batch = next(iter(train_loader))

optimizer.zero_grad()

out = model(
    batch.x,
    batch.edge_index,
    batch.batch
)

loss = criterion(out, batch.y)

loss.backward()

optimizer.step()

print("Loss:", loss.item())
```

```python
def train_one_epoch(model, loader, optimizer, criterion):

    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for batch in loader:

        optimizer.zero_grad()

        out = model(
            batch.x,
            batch.edge_index,
            batch.batch
        )

        loss = criterion(out, batch.y)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * batch.num_graphs

        predictions = out.argmax(dim=1)

        correct += (predictions == batch.y).sum().item()
        total += batch.num_graphs

    average_loss = total_loss / total
    accuracy = correct / total

    return average_loss, accuracy
```

```python
def evaluate(model, loader, criterion):

    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():

        for batch in loader:

            out = model(
                batch.x,
                batch.edge_index,
                batch.batch
            )

            loss = criterion(out, batch.y)

            total_loss += loss.item() * batch.num_graphs

            predictions = out.argmax(dim=1)

            correct += (predictions == batch.y).sum().item()
            total += batch.num_graphs

    average_loss = total_loss / total
    accuracy = correct / total

    return average_loss, accuracy
```

```python
EPOCHS = 20

for epoch in range(1, EPOCHS + 1):

    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion
    )

    val_loss, val_acc = evaluate(
        model,
        val_loader,
        criterion
    )

    print(
        f"Epoch {epoch:02d} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f}"
    )
```

```python
def create_fully_connected_edge_index(num_nodes):

    edges = []

    for i in range(num_nodes):
        for j in range(num_nodes):
            edges.append([i, j])

    edge_index = torch.tensor(
        edges,
        dtype=torch.long
    ).t().contiguous()

    return edge_index
```

```python
def prepare_application_features(
    df,
    service_cols,
    normal_df
):

    # -----------------------------
    # Raw latency
    # -----------------------------
    raw = df[service_cols].astype(float)

    # -----------------------------
    # Normal baseline
    # -----------------------------
    normal_baseline = (
        normal_df[service_cols]
        .astype(float)
        .median()
    )

    # Avoid division by zero
    normal_baseline = normal_baseline.replace(
        0,
        1e-8
    )

    # -----------------------------
    # Ratio to normal
    # -----------------------------
    ratio = raw / normal_baseline

    # -----------------------------
    # Rank within each trace
    # -----------------------------
    rank = raw.rank(
        axis=1,
        method="average",
        ascending=False
    )

    # -----------------------------
    # Contribution
    # -----------------------------
    row_sum = raw.sum(axis=1).replace(
        0,
        1e-8
    )

    contribution = raw.div(
        row_sum,
        axis=0
    )

    return (
        raw,
        ratio,
        rank,
        contribution,
        normal_baseline
    )
```

```python
class ServiceGuardGAT(nn.Module):

    def __init__(
        self,
        num_classes,
        hidden_channels=32,
        heads=4
    ):
        super().__init__()

        self.gat1 = GATConv(
            in_channels=4,
            out_channels=hidden_channels,
            heads=heads
        )

        self.gat2 = GATConv(
            in_channels=hidden_channels * heads,
            out_channels=hidden_channels,
            heads=heads
        )

        self.classifier = nn.Linear(
            hidden_channels * heads,
            num_classes
        )

    def forward(self, x, edge_index, batch):

        x = self.gat1(x, edge_index)
        x = torch.relu(x)

        x = self.gat2(x, edge_index)
        x = torch.relu(x)

        x = global_mean_pool(x, batch)

        x = self.classifier(x)

        return x
```

```python
def predict_trace(
    trace,
    app
):
    
    if app not in metadata:
        raise ValueError(
            f"Unknown application: {app}"
        )

    # -----------------------------
    # Application information
    # -----------------------------
    
    service_cols = metadata[app][
        "service_names"
    ]

    num_nodes = metadata[app][
        "num_nodes"
    ]

    # -----------------------------
    # Make sure required services
    # exist
    # -----------------------------
    
    missing = [
        col
        for col in service_cols
        if col not in trace.columns
    ]

    if missing:
        raise ValueError(
            f"Missing service columns: {missing}"
        )

    # -----------------------------
    # Extract raw features
    # -----------------------------
    
    raw = trace[service_cols].astype(float)

    # -----------------------------
    # Load normal baseline
    # -----------------------------
    
    baseline = application_features[app][
        "baseline"
    ]

    baseline = baseline.replace(
        0,
        1e-8
    )

    # -----------------------------
    # Ratio
    # -----------------------------
    
    ratio = raw / baseline

    # -----------------------------
    # Rank
    # -----------------------------
    
    rank = raw.rank(
        axis=1,
        method="average",
        ascending=False
    )

    # -----------------------------
    # Contribution
    # -----------------------------
    
    row_sum = raw.sum(
        axis=1
    ).replace(
        0,
        1e-8
    )

    contribution = raw.div(
        row_sum,
        axis=0
    )

    # -----------------------------
    # Combine features
    # -----------------------------
    
    features = pd.concat(
        [
            raw,
            ratio.add_suffix("_ratio"),
            rank.add_suffix("_rank"),
            contribution.add_suffix("_contribution")
        ],
        axis=1
    )

    # -----------------------------
    # Load correct scaler
    # -----------------------------
    
    scaler = application_scalers[app]

    features_scaled = scaler.transform(
        features
    )

    # -----------------------------
    # Convert to graph
    # -----------------------------
    
    x = torch.tensor(
        features_scaled[0],
        dtype=torch.float32,
        device=device
    ).view(
        num_nodes,
        4
    )

    edge_index = application_edges[
        app
    ].to(device)

    batch = torch.zeros(
        num_nodes,
        dtype=torch.long,
        device=device
    )

    # -----------------------------
    # Predict
    # -----------------------------
    
    model = final_models[app]

    model.eval()

    with torch.no_grad():

        output = model(
            x,
            edge_index,
            batch
        )

        probabilities = torch.softmax(
            output,
            dim=1
        )[0]

        predicted_class = (
            probabilities.argmax().item()
        )

    # -----------------------------
    # Top-5 predictions
    # -----------------------------
    
    top_probabilities, top_classes = (
        torch.topk(
            probabilities,
            k=min(5, len(probabilities))
        )
    )

    top_predictions = []

    for cls, prob in zip(
        top_classes.cpu().tolist(),
        top_probabilities.cpu().tolist()
    ):

        top_predictions.append({
            "class": cls,
            "fault_name": (
                "No interference"
                if cls == 0
                else service_cols[cls - 1]
                if cls - 1 < len(service_cols)
                else str(cls)
            ),
            "probability": prob * 100
        })

    return (
        predicted_class,
        probabilities.cpu(),
        top_predictions
    )
```

```python
def explain_trace(trace, app, top_k=5):

    service_cols = metadata[app]["service_names"]

    features = application_features[app]

    baseline = features["baseline"].replace(
        0,
        1e-8
    )

    raw = trace[service_cols].astype(float)

    ratio = raw / baseline

    row_sum = raw.sum(axis=1).replace(
        0,
        1e-8
    )

    contribution = raw.div(
        row_sum,
        axis=0
    )

    explanation = pd.DataFrame({
        "service": service_cols,
        "raw_latency": raw.iloc[0].values,
        "normal_latency": baseline.values,
        "ratio_to_normal": ratio.iloc[0].values,
        "contribution": contribution.iloc[0].values
    })

    # Higher ratio = more anomalous
    explanation["anomaly_score"] = (
        explanation["ratio_to_normal"]
        .clip(lower=0)
    )

    explanation = explanation.sort_values(
        "anomaly_score",
        ascending=False
    )

    return explanation.head(top_k)
```

```python
def service_diagnostic_report(
    trace,
    app,
    top_k=5
):

    service_cols = metadata[
        app
    ]["service_names"]

    baseline = application_features[
        app
    ]["baseline"].replace(
        0,
        1e-8
    )

    raw = trace[
        service_cols
    ].astype(float)

    ratios = (
        raw.iloc[0].values /
        baseline.values
    )

    total_latency = (
        raw.iloc[0].sum()
    )

    contributions = (
        raw.iloc[0].values /
        total_latency
    )

    report = pd.DataFrame({
        "service": service_cols,
        "observed_latency": raw.iloc[0].values,
        "normal_latency": baseline.values,
        "slowdown_x": ratios,
        "contribution_pct": contributions * 100
    })

    def severity(ratio):

        if ratio >= 10:
            return "CRITICAL"

        elif ratio >= 5:
            return "HIGH"

        elif ratio >= 2:
            return "MEDIUM"

        elif ratio >= 1.2:
            return "LOW"

        else:
            return "NORMAL"

    report["severity"] = (
        report["slowdown_x"]
        .apply(severity)
    )

    report = report.sort_values(
        "slowdown_x",
        ascending=False
    )

    return report.head(top_k)
```

```python
def serviceguard_predict(
    trace,
    app,
    top_k=5
):

    # -----------------------------
    # Fault prediction
    # -----------------------------

    (
        predicted_class,
        probabilities,
        top_predictions
    ) = predict_trace(
        trace,
        app
    )

    confidence = (
        probabilities[predicted_class].item()
        * 100
    )

    # -----------------------------
    # Service diagnosis
    # -----------------------------

    diagnostic = service_diagnostic_report(
        trace,
        app,
        top_k=top_k
    )

    # -----------------------------
    # Predicted fault name
    # -----------------------------

    if predicted_class == 0:

        fault_name = "No interference"

    else:

        fault_name = metadata[
            app
        ]["service_names"][
            predicted_class - 1
        ]

    # -----------------------------
    # Final result
    # -----------------------------

    return {
        "application": app,
        "predicted_class": predicted_class,
        "predicted_fault": fault_name,
        "confidence": confidence,
        "top_predictions": top_predictions,
        "diagnostic": diagnostic
    }
```
